#### Pandas-DataFrame And Series
Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
import numpy as np
```

#### Series

- it is a one dimensional array that can hold any data type. It is similar to a column in a spreadsheet or a database table. Each element in a Series has an associated index, which allows for easy access and manipulation of the data.

```python
data = [1, 2, 3, 4, 5]
series = pd.Series(data)
print(f"Series:\n{series}\n")
print(type(series))  # Output: <class 'pandas.core.series.Series'>
```

```python
# create a Series with dictionary

data = {'a': 1, 'b': 2, 'c': 3}
series_dict = pd.Series(data)
print(f"Series:\n{series_dict}\n")
```

```python
data  = [10,20,30]
index = ['a','b','c']
series_with_index = pd.Series(data,index= index)
print(f"Series with index:\n{series_with_index}\n")
```

#### Data Frame

- A DataFrame in Pandas is a two-dimensional, labeled data structure that stores data in rows and columns, similar to a table in Excel.

```python
# Data frame from dictionary of lists

data = {
    'Name': ['Krish','John','Jack'],
    'Age': [25, 30, 45],
    'City': ['Banglore', 'New York', 'Florida']
}

df =pd.DataFrame(data)
print(f"DataFrame:\n{df}\n")  
print(type(df))  # Output: <class 'pandas.core.frame.DataFrame'>  
```

```python
# DAta frame from list of dictionaries

data = [
    {'Name': 'Krish', 'Age': 32, 'City': 'Banglore'},
    {'Name': 'John', 'Age': 34, 'City': 'Banglore'},
    {'Name': 'Jack', 'Age': 32, 'City': 'Banglore'},
    {'Name': 'Bappy', 'Age': 32, 'City': 'Banglore'},
]
df = pd.DataFrame(data)
print(f"DataFrame:\n{df}\n")
```

In Pandas, inplace is a parameter that determines whether a modification is applied directly to the original DataFrame or whether a new DataFrame is returned.
- inplace=True: Modifies the original DataFrame directly.
- inplace=False: Does not modify the original DataFrame directly; this is the default for many Pandas methods.

```python
df.drop('Salary', axis=1, inplace=True)  # Dropping a column
df
```

```python
# Adding age to the columns of the DataFrame

df['Age'] += 1
df
```

```python
df.drop(0, axis=0, inplace=True)  # Dropping a row
df
```
