```python
import numpy as np
```

#### Level 1

```python
# Create a NumPy array containing the numbers 10, 20, 30, 40, 50. Print the array, its shape, its size and its data type.

arr = np.array([10,20,30,40,50])
print("Array:", arr)
print("Shape:", arr.shape)
print("Size:",arr.size)
print("DaTa Type:",arr.dtype)
```

```python
# Given arr = np.array([12, 25, 37, 49, 56, 68]), extract the first element, last element and third element.

arr = np.array([12, 25, 37, 49, 56, 68])

print(arr[0])
print(arr[-1])
print(arr[2])
```

```python
# Using the same array, extract the first four elements, the last three elements, and every second element.Show hint

print(arr[0:4])
print(arr[len(arr)-3:])
print(arr[::2])
```

```python
# Create a 3×3 array containing numbers 1 through 9. Extract the second row, third column and the element 8.

arr = np.arange(1,10).reshape(3,3)
arr
```

```python
print(arr[1])
print(arr[:,2])
print(arr[2,1])
```

```python
# Create an array containing numbers 1 through 12 and reshape it into a 3×4 matrix. Then reshape it into a 2×6 matrix.

arr = np.arange(1,13).reshape(3,4)
print(f"Array on first reshape: {arr}",end= "\n")

arr.reshape(2,6)
print(f"Array on Second reshape: {arr}")
```

```python
# Create an array of 10 zeros, an array of 5 ones, and an array containing even numbers from 2 to 20.

arr1 = np.zeros(10)
arr1
```

```python
arr2 = np.ones(5)
arr2
```

```python
# Create an array of integers and convert it into float values using astype(). Print the new data type.

arr = np.arange(1,10)
print(f"Before:{arr.dtype}")
arr2 = arr.astype(float)
print(f"After:{arr2.dtype}")
```

```python
# Given a = np.array([2, 4, 6, 8]) and b = np.array([1, 2, 3, 4]), calculate their sum, difference, product and element-wise division.

a = np.array([2, 4, 6, 8])
b = np.array([1, 2, 3, 4])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
# Given arr = np.array([10, 15, 20, 25, 30, 35, 40]), extract all values greater than 25.

arr = np.array([10, 15, 20, 25, 30, 35, 40])
arr[arr[:] > 25]
```

```python
# Given data = np.array([18, 21, 17, 25, 16, 22, 19]), count how many values are below 19.

data = np.array([18, 21, 17, 25, 16, 22, 19])
len(data[data[:]< 19])
```

#### Level 2

```python
# Convert a 1D array of 24 elements into a 4×6 matrix.

arr = np.arange(0,24)
print(f"1D array: {arr}",end= "\n")

arr = arr.reshape(4,6)
print(f"2D array: \n {arr}",end= "\n")
```

```python
# Create a 2×3 matrix and convert it into a 3×2 matrix using transpose.

arr = np.arange(0,6).reshape(2,3)

print(f"Array:\n {arr}",end="\n")

arr_transpose = arr.T
print(f"Transposed Array:\n {arr_transpose}",end="\n")
```

```python
# Convert a 3×3 matrix into a 1D array.

arr = np.arange(0,9).reshape(3,3)
arr
```

```python
arr_flat = arr.flatten()
print(f"Flat matrix: {arr_flat}")
```
