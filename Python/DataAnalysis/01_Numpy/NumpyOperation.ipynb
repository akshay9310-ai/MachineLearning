#### Broadcasting

- Broadcasting in NumPy is a mechanism that allows NumPy to perform arithmetic operations on arrays of different shapes without explicitly creating copies of the smaller array.

```python
l = [10,20,30,40]
arr = np.array(l)
arr
```

```python
arr + 10
```

```python
arr2 = np.arange(1,26).reshape(5,5)
arr2 + 10
```

#### Deep and Shallow Copy

- A shallow copy creates a new object, but the elements inside it are not copied recursively. Nested objects are shared between the original and copied objects.

- A deep copy creates a new object and recursively copies its nested objects, so changes to nested mutable objects in the copy do not affect the original.

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice *= 10
slice
```

- One important thing when we slice something out of an array the sliced part creates it's own space in the memory and if we make changes in the sliced portion they are not reflected in the main array

- But if we initialise other array as ( arr2 = arr1) then both arr1 and arr2 point to the same location in the memory and change in one reflects in other

```python
a
```

```python
# Shallow Copy
b = a.copy()
```

```python
b[0] = 99
print(b)
```

```python
print(a)
```

#### Advance Array Manipulation

##### Stacking Array

- Array stacking in NumPy is the process of joining two or more arrays along a specified axis to form a single array.

- NumPy provides functions such as np.vstack() and np.hstack() to stack arrays vertically and horizontally, respectively.

- You cannot stack two matrix, you can only stack two arrays or vectors.

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])

np.vstack((a,b))
```

```python
np.hstack((a,b))
```

```python
np.column_stack((a,b))
```

##### Spliting Array

- Array splitting in NumPy is the process of dividing a single array into multiple smaller arrays using functions such as np.split(), np.array_split(), np.hsplit(), and np.vsplit().

- Arrays can only be splitted into equal divisions.

```python
c = np.arange(16).reshape(4,4)
c
```

```python
np.hsplit(c,2)
```

```python
np.hsplit(c,4)
```

```python
a = np.vsplit(c,2)
a
```

```python
for i in a:
    print(i)
```
