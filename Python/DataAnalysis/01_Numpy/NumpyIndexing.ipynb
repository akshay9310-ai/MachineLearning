#### Numpy Indexing and Slicing of Vectors

- A vector in NumPy is a one-dimensional array used to store a collection of elements of the same data type.

```python
arr =np.arange(11,21)
arr
```

```python
arr[6]
```

```python
arr[:5]
```

```python
arr_slice = arr[3::2]
arr_slice
```

#### Numpy Indexing and Slicing of Matrix

- Indexing is the process of accessing an individual element of a sequence (such as a string, list, or tuple) using its index or position.

- Slicing is the process of accessing a range of elements from a sequence by specifying a start index, an end index, and optionally a step.

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0,4]
```

```python
arr[5,4]
```

```python
slice_1 = arr[0:2,1:3]
slice_1
```

```python
slice_2 = arr[3:6,3:5]
slice_2
```

```python
# Selecting all the rows and 3rd column
slice_3 = arr[:,2]
slice_3
```
